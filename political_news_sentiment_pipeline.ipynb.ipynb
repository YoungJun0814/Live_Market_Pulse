```python
import feedparser
import pandas as pd
from datetime import datetime
import time
from textblob import TextBlob

# 🔗 RSS SOURCE (Google News — real-time enough)
RSS_URL = "https://news.google.com/rss/search?q=Trump+Netanyahu+Iran&hl=en-US&gl=US&ceid=US:en"

KEYWORDS = ["Trump", "Netanyahu", "Iran"]

# 🟢 1. FETCH DATA
def fetch_rss():
    feed = feedparser.parse(RSS_URL)

    articles = []

    for entry in feed.entries:
        articles.append({
            "title": entry.title,
            "link": entry.link,
            "published": entry.published,
            "published_dt": datetime(*entry.published_parsed[:6])
        })

    df = pd.DataFrame(articles)
    return df


# 🟢 2. FILTER POLITICAL SIGNALS
def filter_keywords(df):
    return df[df["title"].str.contains("|".join(KEYWORDS), case=False, na=False)]


# 🟢 3. FORMAT TIMESTAMP
def format_time(df):
    df["published_dt"] = pd.to_datetime(df["published_dt"])
    df["formatted_time"] = df["published_dt"].dt.strftime("%m/%d/%Y %H:%M")
    return df


# 🟢 4. SENTIMENT (FAST VERSION)
def get_sentiment(text):
    score = TextBlob(text).sentiment.polarity  # -1 to 1
    return int(score * 100)  # convert to -100 to 100


def add_sentiment(df):
    df["sentiment_score"] = df["title"].apply(get_sentiment)
    return df


# 🟢 5. MAP TO YOUR JSON FORMAT
def to_signal_json(df):
    signals = []

    for _, row in df.iterrows():
        signals.append({
            "source": "politician",
            "sentiment_score": row["sentiment_score"],
            "confidence": 70,  # placeholder
            "signal": row["title"],
            "author": "unknown",
            "platform": "rss",
            "tier": "major",
            "event_location": {
                "city": None,
                "country": None,
                "lat": None,
                "lon": None
            },
            "timestamp": row["published_dt"].isoformat()
        })

    return signals


# 🟢 6. MAIN LOOP (REAL-TIME SIMULATION)
def run_pipeline():
    df = fetch_rss()

    if df.empty:
        print("⚠️ No data fetched")
        return

    df = filter_keywords(df)

    if df.empty:
        print("⚠️ No matching political news")
        return

    df = format_time(df)
    df = add_sentiment(df)

    print("\n🟢 LIVE POLITICAL SIGNALS:\n")
    print(df[["title", "formatted_time", "sentiment_score"]].head())

    signals = to_signal_json(df)

    print("\n📡 JSON OUTPUT (for Central Hub):\n")
    for s in signals[:3]:
        print(s)


# 🔁 LOOP EVERY 60 SECONDS
while True:
    run_pipeline()
    time.sleep(60)
```
